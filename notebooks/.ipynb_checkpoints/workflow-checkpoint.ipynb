```python
import kagglehub
from pathlib import Path

# Download latest version
path = kagglehub.dataset_download("ahmedsorour1/mri-for-brain-tumor-with-bounding-boxes")

DATASET_PATH = Path(path)
```

```python
from pathlib import Path

def count_files_recursively(current_path: Path, root_path: Path):
    if not current_path.exists() or not current_path.is_dir():
        return

    items = list(current_path.iterdir())
    files = [item for item in items if item.is_file()]
    
    if len(files) > 0:
        relative_path = current_path.relative_to(root_path.parent)
        print(f"{relative_path}: {len(files)}")
    
    for item in items:
        if item.is_dir():
            count_files_recursively(item, root_path)

count_files_recursively(DATASET_PATH, Path(path))
```
